## Περιγραφή Reservoir Sampling

Ο τροποποιημένος αλγόριθμος Reservoir Sampling θα κρατάει χώρο για $Κ$ αντικείμενα (reservoir size) και ακόμη έναν μετρητή για να καταγράφει πόσα δεδομένα έχουν εμφανιστεί.

Αρχικά τοποθετεί τα πρώτα K αντικείμενα στο reservoir τα οποία αποτελούν το προσωρινό δείγμα.

Έπειτα για κάθε επόμενο αντικείμενο $(Κ+1, Κ+2, ...)$ ο αλγόριθμος το εξετάζει με πιθανότητα $\frac{K}{n}$, όπου $n$ είναι η σειρά που εμφανίζονται σύμφωνα με τον μετρητή. Αν επιλεγεί τότε αντικαθιστά τυχαία ένα από τα αποθηκευμένα $K$ αντικείμενα με πιθανότητα $\frac{1}{K}$ για το καθένα.

Ενημερώνει το μετρητή και συνεχίζει με το επόμενο αντικείμενο.

Έτσι ο αλγόριθμος επεξεργάζεται μόνο μια φορά κάθε αντικείμενο, χρησιμοποιεί χώρο $O(K)$ (reservoir size + counter) και δεν γνωρίζει εκ των προτέρων πόσα συνολικά δεδομένα περιέχει το ρεύμα $(N)$.

Όταν ολοκληρωθεί η ροή, επιστρέφει το reservoir με τα $K$ αντικείμενα που έχουν επιλεχθεί ομοιόμορφα τυχαία από το σύνολο.
