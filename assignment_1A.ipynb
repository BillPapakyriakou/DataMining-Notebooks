## Περιγραφή Reservoir Sampling

Ο τροποποιημένος αλγόριθμος Reservoir Sampling θα κρατάει χώρο για $Κ$ αντικείμενα (reservoir size) και ακόμη έναν μετρητή για να καταγράφει πόσα δεδομένα έχουν εμφανιστεί.

Αρχικά τοποθετεί τα πρώτα K αντικείμενα στο reservoir τα οποία αποτελούν το προσωρινό δείγμα.

Έπειτα για κάθε επόμενο αντικείμενο $(Κ+1, Κ+2, ...)$ ο αλγόριθμος το εξετάζει με πιθανότητα $\frac{K}{n}$, όπου $n$ είναι η σειρά που εμφανίζονται σύμφωνα με τον μετρητή. Αν επιλεγεί τότε αντικαθιστά τυχαία ένα από τα αποθηκευμένα $K$ αντικείμενα με πιθανότητα $\frac{1}{K}$ για το καθένα.

Ενημερώνει το μετρητή και συνεχίζει με το επόμενο αντικείμενο.

Έτσι ο αλγόριθμος επεξεργάζεται μόνο μια φορά κάθε αντικείμενο, χρησιμοποιεί χώρο $O(K)$ (reservoir size + counter) και δεν γνωρίζει εκ των προτέρων πόσα συνολικά δεδομένα περιέχει το ρεύμα $(N)$.

Όταν ολοκληρωθεί η ροή, επιστρέφει το reservoir με τα $K$ αντικείμενα που έχουν επιλεχθεί ομοιόμορφα τυχαία από το σύνολο.   






## Απόδειξη ότι ο αλγόριθμος παράγει ένα ομοιόμορφα τυχαίο δείγμα

Θα δείξουμε με μαθηματική επαγωγή ότι μετά την επεξεργασία των πρώτων $i$ στοιχείων του ρεύματος,
κάθε στοιχείο έχει πιθανότητα $\frac{K}{i}$ να βρίσκεται στο δείγμα $T_i$ (reservoir).

---

### Βάση $(i = K)$

Τα πρώτα $K$ στοιχεία τοποθετούνται στο reservoir, άρα ισχύει:

$Pr(x_i \in T_i) = \frac{K}{i} = 1$

---

### Επαγωγική υπόθεση
Υποθέτουμε ότι μετά την επεξεργασία των πρώτων $i$ στοιχείων, κάθε στοιχείο έχει πιθανότητα $\frac{K}{i}$ για να ανήκει στο δείγμα $T_i$.

---

### Επαγωγικό βήμα $(i$ $-$> $i+1)$
Ο αλγόριθμος επιλέγει το $(i+1)-$οστό στοιχείο με πιθανότητα $\frac{K}{i+1}$.
Αν επιλεγεί, τότε για κάθε ένα από τα $K$ στοιχεία του $T_i$ έχει πιθανότητα $\frac{1}{K}$ να αντικατασταθεί.

Άρα η πιθανότητα ένα συγκεκριμένο στοιχείο να αντικατασταθεί είναι:
$\frac{K}{i+1} \cdot \frac{1}{K} = \frac{1}{i+1}$

Η πιθανότητα να μην αντικατασταθεί είναι:
$1 - \frac{1}{i+1} = \frac{i}{i+1}$

Επομένως η πιθανότητα το στοιχείο να παραμείνει στο reservoir μετά το νέο βήμα είναι:
$Pr(x \in T_{i+1}) = Pr(x \in T_{i}) \cdot Pr($δεν αντικαταστάθηκε$) \ = \frac{K}{i} \cdot \frac{i}{i+1} = \frac{K}{i+1}$

Το νέο στοιχείο επίσης επιλέγεται με πιθανότητα $\frac{K}{i+1}$.

---
### Συμπέρασμα
Άρα όλα τα στοιχεία έχουν πιθανότητα $\frac{K}{i+1}$ να βρίσκονται στο reservoir. Με επαγωγή προκύπτει ότι για κάθε $i \le N$:

$Pr(x_i \in T_N) = \frac{K}{N}$

Δηλαδή ο αλγόριθμος παράγει ομοιόμορφα τυχαίο δείγμα μεγέθους Κ.
