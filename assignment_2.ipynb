### Απόδειξη:
Για κάθε σημείο $a_i$, η προβολή του στο $u$ είναι:
$$(a_i\cdot u)\,u.$$

Η  κάθετη απόσταση του $a_i$ από την προβολή του είναι:
$$\lVert a_i - (a_i\cdot u)u\rVert.$$

Το συνολικό άθροισμα τετραγώνων των αποστάσεων είναι:
$$
F(u)=\sum_{i=1}^n \lVert a_i - (a_i\cdot u)u\rVert^2.
$$

Έχουμε:
$$
\|a_i - (a_i\cdot u)u\|^2
= (a_i - (a_i\cdot u)u)^\top (a_i - (a_i\cdot u)u) =
$$

$$
a_i^\top a_i
- (a_i\cdot u)\, a_i^\top u
- (a_i\cdot u)\, u^\top a_i
+ (a_i\cdot u)^2\, u^\top u
$$

Χρησιμοποιούμε ότι $a_i^\top u = u^\top a_i = a_i\cdot u$ και $u^\top u = 1$:
$$
= \|a_i\|^2
- (a_i\cdot u)^2
- (a_i\cdot u)^2
+ (a_i\cdot u)^2 =
$$

$$
\|a_i\|^2 - (a_i\cdot u)^2
.
$$

Άρα:
$$
F(u)=\sum_{i=1}^n \lVert a_i\rVert^2
     - \sum_{i=1}^n (a_i\cdot u)^2.
$$

Το πρώτο άθροισμα είναι ανεξάρτητο του $u$ (άρα σταθερό), επομένως:

$$
\text{ελαχιστοποίηση }F(u)
\quad\Longleftrightarrow\quad
\text{μεγιστοποίηση }G(u),
$$
όπου θέτουμε:
$$
G(u) = \sum_{i=1}^n (a_i\cdot u)^2.
$$

Παρατηρούμε ότι:
$$
G(u) = \lVert A u\rVert^2 = u^\top (A^\top A) u,
$$
με περιορισμό:
$$
\lVert u\rVert^2 = u^\top u = 1.
$$

---

Χρησιμοποιώντας τη μέθοδο πολλαπλασιαστών Lagrange, ορίζουμε:
$$
\mathcal{L}(u,\lambda)
   = u^\top A^\top A u - \lambda (u^\top u - 1).
$$

Παίρνουμε παράγωγο ως προς $u$ ίση με το μηδέν:
$$
\nabla_u \mathcal{L}
= 2A^\top A u - 2\lambda u = 0,
$$
οπότε:
$$
A^\top A u = \lambda u.
$$

Άρα $u$ πρέπει να είναι ιδιοδιάνυσμα του $A^\top A$.


$$
G(u) = u^\top A^\top A u
     = u^\top (\lambda u)
     = \lambda \, (u^\top u)
     = \lambda,
$$
αφού $u^\top u = 1$.

$\rightarrow$ Άρα ψάχνουμε το ιδιοδιάνυσμα με τη μεγαλύτερη ιδιοτιμή $\lambda_{\max}.$

---

Από τις διαφάνειες έχουμε:

- Oι ιδιοτιμές του $A^\top A$ είναι $\lambda_i=\sigma_i^2$

- Tα ιδιοδιανύσματά του $A^\top A$ είναι οι δεξιοί singular vectors $v_i$.

- O πρώτος δεξιός singular vector $v_1$ αντιστοιχεί στη μεγαλύτερη ιδιοτιμή $\sigma_1^2$ και μεγιστοποιεί το $G(u)$.

(συγκεκριμένα στο dimensionality-reduction.pdf, διαφάνεια 24 και διαφάνεια 36)

Ο πρώτος δεξιός singular vector $v_1$ ελαχιστοποιεί το άθροισμα των τετραγώνων των αποστάσεων των $a_i$ από τις προβολές τους πάνω στο διάνυσμα $v$.

## **Ερώτηση 2**

##Α) Maximum Likelihood Estimation για να βρούμε τον εκθέτη $α$ στην power-law κατανομή

Δίνεται:

$$
P(X = x) = (a - 1)x^{-a}.
$$

και δείγμα:

$$
X = \{x_1, x_2, \dots, x_n\}.
$$

---

Η συνάρτηση πιθανοφάνειας είναι:

$$
L(a) = \prod_{i=1}^n (a - 1)x_i^{-a}
      = (a - 1)^n \prod_{i=1}^n x_i^{-a}.
$$


Παίρνουμε το log-likelihood (λογαριθμίζουμε):

$$
\ell(a) = \log L(a)
        = n\log(a - 1)
        - a \sum_{i=1}^n \log x_i.
$$



Παίρνουμε την παράγωγο ως προς $a$ ίση με το μηδέν:

$$
\frac{d\ell}{da}
= \frac{n}{a - 1}
 - \sum_{i=1}^n \log x_i = 0 \Leftrightarrow
$$

$$
\frac{n}{a - 1}
= \sum_{i=1}^n \log x_i.
$$

Λύνουμε ως προς $a$:

$$
a - 1
= \frac{n}{\sum_{i=1}^n \log x_i} \Leftrightarrow
$$


$$
a
= 1 + \frac{n}{\sum_{i=1}^n \log x_i}
.
$$

---

Παίρνουμε για επαλήθευση τη δεύτερη παράγωγο:

$$
\frac{d^2 \ell}{da^2}
= -\frac{n}{(a - 1)^2}
< 0,
$$

Άρα το σημείο είναι πράγματι μέγιστο.

---

### Συμπέρασμα

Ο εκθέτης $α$ της power-law κατανομής εκτιμάται με Maximum Likelihood Estimation σε:

$$
\boxed{
a = 1 + \frac{n}{\sum_{i=1}^n \log x_i}
}
$$

---
##B) EM για μείγμα δύο power–law κατανομών

Οι power-law κατανομές είναι:
$$
p(x \mid L_k) = (a_k - 1)x^{-a_k},
\qquad k = 1,2.
$$
με πιθανότητες μίξης $\pi_1, \pi_2$ όπου $$\pi_1 + \pi_2 = 1$$

---

Από τις σημειώσεις του Άρη Αναγνωστόπουλου (σελίδα 9) έχουμε:

$$
\gamma_{ik} = P(L_k \mid x_i)
=
\frac{\pi_k\, p(x_i \mid L_k)}
     {\sum_{j=1}^2 \pi_j\, p(x_i \mid L_j)}.
$$

όπου $\gamma_{ik}$ : πιθανότητα $x_i$ να προέρχεται από $L_k$ power-law

Έχουμε κατανομή power-law:
$$
p(x_i \mid L_k) = (a_k - 1)x_i^{-a_k},
$$
άρα:
$$
\gamma_{ik}
=
\frac{\pi_k (a_k - 1)x_i^{-a_k}}
     {\sum_{j=1}^2 \pi_j (a_j - 1)x_i^{-a_j}}.
$$

---

Εύρεση των $\pi_k$

Όπως και στο παράδειγμα με τις Gaussian:

$$
\pi_k^{\text{new}}
=
\frac{1}{n}\sum_{i=1}^n \gamma_{ik},
\qquad k = 1,2.
$$

---

Εύρεση των εκθετών $a_k$

Για power-law, στο ερώτημα Α της ερώτησης 2 ο MLE έδωσε:
$$
a = 1 + \frac{n}{\sum_{i=1}^n \log x_i}.
$$

Στο μίγμα δύο κατανομών, κάθε σημείο δεν ανήκει πλήρως σε μια κατανομή $L_k$,  
αλλά με βάρος $\gamma_{ik}$.  

Άρα στο M–βήμα κάνουμε weighted MLE:

- το $n$ αντικαθίσταται με από το άθροισμα των πιθανοτήτων ότι τα σημεία ανήκουν στην $L_k$
  $$
  n \;\to\; \sum_{i=1}^n \gamma_{ik},
  $$

- το $\sum \log x_i$ αντικαθίσταται από weighted άθροισμα logs
  $$
  \sum_{i=1}^n \log x_i
  \;\to\;
  \sum_{i=1}^n \gamma_{ik} \log x_i.
  $$

με αντικατάσταση στη συνάρτηση από το ερώτημα Α καταλήγουμε:

$$
a_k^{\text{new}}
=
1
+
\frac{\sum_{i=1}^n \gamma_{ik}}
     {\sum_{i=1}^n \gamma_{ik} \log x_i},
\qquad k = 1,2.
$$

δηλαδή εφαρμόζουμε τον τύπο MLE της power-law αλλά σε σημεία με βάρη $\gamma_{ik}$

---

### Συμπέρασμα

Για να βρούμε τις παραμέτρους $\pi_1, \pi_2, a_1, a_2$:

$$
\boxed{
\pi_k^{\text{new}} = \frac{1}{n}\sum_{i=1}^n \gamma_{ik}
,\qquad
a_k^{\text{new}} = 1 +
\frac{\sum_{i=1}^n \gamma_{ik}}
     {\sum_{i=1}^n \gamma_{ik}\log x_i},
\quad k = 1,2.
}
$$

## Βήμα 1:

```python
import pandas as pd
from sklearn.metrics import mean_squared_error
import numpy as np

# φόρτωση train και test dataset
train_dataframe = pd.read_csv("sample_data/data_train.csv")
test_dataframe  = pd.read_csv("sample_data/data_test.csv")

# στήλες αρχείων ["uid", "mid", "rating"]
```

###**User Average**

```python
# μέσος όρος rating κάθε χρήστη (training δεδομένα)
user_means = train_dataframe.groupby("uid")["rating"].mean()

# κάνει join στο uid και προσθέτει για κάθε χρήστη το user-average στη στήλη rating_pred
test_user_average = test_dataframe.merge(user_means, on="uid", how="left", suffixes=("", "_pred"))

# rmse μεταξύ train και test data
mse = mean_squared_error(test_user_average["rating"], test_user_average["rating_pred"])
rmse_user_average = np.sqrt(mse)

print("RMSE (User Average):", rmse_user_average)
```

###**Item Average**

```python
# μέσος όρος rating κάθε ταινίας (training δεδομένα)
item_means = train_dataframe.groupby("mid")["rating"].mean()

# κάνει join στο mid και προσθέτει για κάθε ταινία το item-average στη στήλη rating_pred
test_item_average = test_dataframe.merge(item_means, on="mid", how="left", suffixes=("", "_pred"))

# υπολογίζουμε το global mean για να αντικαταστήσουμε της NaN γραμμές με αυτό
global_mean = train_dataframe["rating"].mean()

# αντικατάσταση NaN γραμμών με global mean
test_item_average["rating_pred"] = test_item_average["rating_pred"].fillna(global_mean)

# rmse μεταξύ train και test data
mse = mean_squared_error(test_item_average["rating"], test_item_average["rating_pred"])
rmse_item_average = np.sqrt(mse)

print("RMSE (Item Average):", rmse_item_average)
```

- Για το user average η πρόβλεψη για κάθε χρήστη είναι ο μέσος όρος των ratings από το training set.

- Για το item average είχαμε ταινίες που δεν υπήρχαν στο training set, γι αυτό για να γίνει η πρόβλεψη κάναμε αντικατάσταση των NaN γραμμών με το μέσο όρο των rating.
