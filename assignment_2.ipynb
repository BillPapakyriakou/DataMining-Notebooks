### Απόδειξη:
Για κάθε σημείο $a_i$, η προβολή του στο $u$ είναι:
$$(a_i\cdot u)\,u.$$

Η  κάθετη απόσταση του $a_i$ από την προβολή του είναι:
$$\lVert a_i - (a_i\cdot u)u\rVert.$$

Το συνολικό άθροισμα τετραγώνων των αποστάσεων είναι:
$$
F(u)=\sum_{i=1}^n \lVert a_i - (a_i\cdot u)u\rVert^2.
$$

Έχουμε:
$$
\|a_i - (a_i\cdot u)u\|^2
= (a_i - (a_i\cdot u)u)^\top (a_i - (a_i\cdot u)u) =
$$

$$
a_i^\top a_i
- (a_i\cdot u)\, a_i^\top u
- (a_i\cdot u)\, u^\top a_i
+ (a_i\cdot u)^2\, u^\top u
$$

Χρησιμοποιούμε ότι $a_i^\top u = u^\top a_i = a_i\cdot u$ και $u^\top u = 1$:
$$
= \|a_i\|^2
- (a_i\cdot u)^2
- (a_i\cdot u)^2
+ (a_i\cdot u)^2 =
$$

$$
\|a_i\|^2 - (a_i\cdot u)^2
.
$$

Άρα:
$$
F(u)=\sum_{i=1}^n \lVert a_i\rVert^2
     - \sum_{i=1}^n (a_i\cdot u)^2.
$$

Το πρώτο άθροισμα είναι ανεξάρτητο του $u$ (άρα σταθερό), επομένως:

$$
\text{ελαχιστοποίηση }F(u)
\quad\Longleftrightarrow\quad
\text{μεγιστοποίηση }G(u),
$$
όπου θέτουμε:
$$
G(u) = \sum_{i=1}^n (a_i\cdot u)^2.
$$

Παρατηρούμε ότι:
$$
G(u) = \lVert A u\rVert^2 = u^\top (A^\top A) u,
$$
με περιορισμό:
$$
\lVert u\rVert^2 = u^\top u = 1.
$$

---

Χρησιμοποιώντας τη μέθοδο πολλαπλασιαστών Lagrange, ορίζουμε:
$$
\mathcal{L}(u,\lambda)
   = u^\top A^\top A u - \lambda (u^\top u - 1).
$$

Παίρνουμε παράγωγο ως προς $u$ ίση με το μηδέν:
$$
\nabla_u \mathcal{L}
= 2A^\top A u - 2\lambda u = 0,
$$
οπότε:
$$
A^\top A u = \lambda u.
$$

Άρα $u$ πρέπει να είναι ιδιοδιάνυσμα του $A^\top A$.


$$
G(u) = u^\top A^\top A u
     = u^\top (\lambda u)
     = \lambda \, (u^\top u)
     = \lambda,
$$
αφού $u^\top u = 1$.

$\rightarrow$ Άρα ψάχνουμε το ιδιοδιάνυσμα με τη μεγαλύτερη ιδιοτιμή $\lambda_{\max}.$

---

Από τις διαφάνειες έχουμε:

- Oι ιδιοτιμές του $A^\top A$ είναι $\lambda_i=\sigma_i^2$

- Tα ιδιοδιανύσματά του $A^\top A$ είναι οι δεξιοί singular vectors $v_i$.

- O πρώτος δεξιός singular vector $v_1$ αντιστοιχεί στη μεγαλύτερη ιδιοτιμή $\sigma_1^2$ και μεγιστοποιεί το $G(u)$.

(συγκεκριμένα στο dimensionality-reduction.pdf, διαφάνεια 24 και διαφάνεια 36)

---

### Συμπέρασμα

Ο πρώτος δεξιός singular vector $v_1$ ελαχιστοποιεί το άθροισμα των τετραγώνων των αποστάσεων των $a_i$ από τις προβολές τους πάνω στο διάνυσμα $v$.

## **Ερώτηση 2**

##Α) Maximum Likelihood Estimation για να βρούμε τον εκθέτη $α$ στην power-law κατανομή

Δίνεται:

$$
P(X = x) = (a - 1)x^{-a}.
$$

και δείγμα:

$$
X = \{x_1, x_2, \dots, x_n\}.
$$

---

Η συνάρτηση πιθανοφάνειας είναι:

$$
L(a) = \prod_{i=1}^n (a - 1)x_i^{-a}
      = (a - 1)^n \prod_{i=1}^n x_i^{-a}.
$$


Παίρνουμε το log-likelihood (λογαριθμίζουμε):

$$
\ell(a) = \log L(a)
        = n\log(a - 1)
        - a \sum_{i=1}^n \log x_i.
$$



Παίρνουμε την παράγωγο ως προς $a$ ίση με το μηδέν:

$$
\frac{d\ell}{da}
= \frac{n}{a - 1}
 - \sum_{i=1}^n \log x_i = 0 \Leftrightarrow
$$

$$
\frac{n}{a - 1}
= \sum_{i=1}^n \log x_i.
$$

Λύνουμε ως προς $a$:

$$
a - 1
= \frac{n}{\sum_{i=1}^n \log x_i} \Leftrightarrow
$$


$$
a
= 1 + \frac{n}{\sum_{i=1}^n \log x_i}
.
$$

---

Παίρνουμε για επαλήθευση τη δεύτερη παράγωγο:

$$
\frac{d^2 \ell}{da^2}
= -\frac{n}{(a - 1)^2}
< 0,
$$

Άρα το σημείο είναι πράγματι μέγιστο.

---

### Συμπέρασμα

Ο εκθέτης $α$ της power-law κατανομής εκτιμάται με Maximum Likelihood Estimation σε:

$$
\boxed{
a = 1 + \frac{n}{\sum_{i=1}^n \log x_i}
}
$$

##B) EM για μείγμα δύο power–law κατανομών

Οι power-law κατανομές είναι:
$$
p(x \mid L_k) = (a_k - 1)x^{-a_k},
\qquad k = 1,2.
$$
με πιθανότητες μίξης $\pi_1, \pi_2$ όπου $$\pi_1 + \pi_2 = 1$$

---

Από τις σημειώσεις του Άρη Αναγνωστόπουλου (σελίδα 9) έχουμε:

$$
\gamma_{ik} = P(L_k \mid x_i)
=
\frac{\pi_k\, p(x_i \mid L_k)}
     {\sum_{j=1}^2 \pi_j\, p(x_i \mid L_j)}.
$$

όπου $\gamma_{ik}$ : πιθανότητα $x_i$ να προέρχεται από $L_k$ power-law

Έχουμε κατανομή power-law:
$$
p(x_i \mid L_k) = (a_k - 1)x_i^{-a_k},
$$
άρα:
$$
\gamma_{ik}
=
\frac{\pi_k (a_k - 1)x_i^{-a_k}}
     {\sum_{j=1}^2 \pi_j (a_j - 1)x_i^{-a_j}}.
$$

---

Εύρεση των $\pi_k$

Όπως και στο παράδειγμα με τις Gaussian:

$$
\pi_k^{\text{new}}
=
\frac{1}{n}\sum_{i=1}^n \gamma_{ik},
\qquad k = 1,2.
$$

---

Εύρεση των εκθετών $a_k$

Το μοντέλο μίξης αποτελείται από δύο power-law κατανομές:

$$
p(x_i \mid L_k) = (a_k - 1)\, x_i^{-a_k}, \qquad k = 1,2.
$$

Εισάγουμε λανθάνουσες μεταβλητές $z_{ik} \in \{0,1\}$ με $z_{i1}+z_{i2}=1$

(αντί για $z_{ik} \in \{I, C\}$ στο παράδειγμα με Gaussian)

Η συνάρτηση πιθανοφάνειας για $(x_i, z_i)$ είναι:

$$
L_c(\Theta)
=
\prod_{i=1}^n
\prod_{k=1}^2
\left[ \pi_k (a_k - 1) x_i^{-a_k} \right]^{z_{ik}}.
$$

Παίρνουμε το log-likelihood (λογαριθμίζουμε):

$$
\ell_c(\Theta)
=
\sum_{i=1}^n \sum_{k=1}^2
z_{ik} \left( \log \pi_k + \log(a_k - 1) - a_k \log x_i \right).
$$


Όπως στις σημειώσεις του Αναγνωστόπουλου με το μίγμα των Gaussian, οι άγνωστες λανθάνουσες ποσότητες εκφράζονται μέσω των πιθανοτήτων ανάθεσης
$\gamma_{ik}$, οπότε η log-likelihood ξανα-γράφεται:


$$
\ell_c(\Theta)
=
\sum_{i=1}^n \sum_{k=1}^2
\gamma_{ik}
\left(
\log \pi_k + \log(a_k - 1) - a_k \log x_i
\right).
$$

Κρατάμε μόνο τους όρους που εξαρτώνται από το $a_k$:

$$
\ell_c(a_k)
=
\sum_{i=1}^n \gamma_{ik} \log(a_k - 1)
-
a_k \sum_{i=1}^n \gamma_{ik} \log x_i.
$$

Παίρνουμε παράγωγο ως προς $a_k$ ίση με το μηδέν:

$$
\frac{d\ell_c}{da_k}
=
\sum_{i=1}^n \frac{\gamma_{ik}}{a_k - 1}
-
\sum_{i=1}^n \gamma_{ik} \log x_i = 0.
$$


$$
\sum_{i=1}^n \frac{\gamma_{ik}}{a_k - 1}
=
\sum_{i=1}^n \gamma_{ik} \log x_i.
$$

Λύνουμε ως προς $a_k$:

$$
\frac{1}{a_k - 1}
=
\frac{\sum_{i=1}^n \gamma_{ik} \log x_i}
     {\sum_{i=1}^n \gamma_{ik}},
$$

$$
a_k - 1
=
\frac{\sum_{i=1}^n \gamma_{ik}}
     {\sum_{i=1}^n \gamma_{ik} \log x_i},
$$


$$
a_k
=
1
+
\frac{\sum_{i=1}^n \gamma_{ik}}
     {\sum_{i=1}^n \gamma_{ik} \log x_i}
.
$$


---

### Συμπέρασμα

Για να βρούμε τις παραμέτρους $\pi_1, \pi_2, a_1, a_2$:

$$
\boxed{
\pi_k^{\text{new}} = \frac{1}{n}\sum_{i=1}^n \gamma_{ik}
,\qquad
a_k^{\text{new}} = 1 +
\frac{\sum_{i=1}^n \gamma_{ik}}
     {\sum_{i=1}^n \gamma_{ik}\log x_i},
\quad k = 1,2.
}
$$

## Βήμα 1:

```python
import pandas as pd
from sklearn.metrics import mean_squared_error
import numpy as np

# φόρτωση train και test dataset
train_dataframe = pd.read_csv("sample_data/data_train.csv")
test_dataframe  = pd.read_csv("sample_data/data_test.csv")

# στήλες αρχείων ["uid", "mid", "rating"]
```

###**User Average**

```python
# μέσος όρος rating κάθε χρήστη (training δεδομένα)
user_means = train_dataframe.groupby("uid")["rating"].mean()

# κάνει join στο uid και προσθέτει για κάθε χρήστη το user-average στη στήλη rating_pred
test_user_average = test_dataframe.merge(user_means, on="uid", how="left", suffixes=("", "_pred"))

# rmse μεταξύ train και test data
mse = mean_squared_error(test_user_average["rating"], test_user_average["rating_pred"])
rmse_user_average = np.sqrt(mse)

print("RMSE (User Average):", rmse_user_average)
```

###**Item Average**

```python
# μέσος όρος rating κάθε ταινίας (training δεδομένα)
item_means = train_dataframe.groupby("mid")["rating"].mean()

# κάνει join στο mid και προσθέτει για κάθε ταινία το item-average στη στήλη rating_pred
test_item_average = test_dataframe.merge(item_means, on="mid", how="left", suffixes=("", "_pred"))

# υπολογίζουμε το global mean για να αντικαταστήσουμε της NaN γραμμές με αυτό
global_mean = train_dataframe["rating"].mean()

# αντικατάσταση NaN γραμμών με global mean
test_item_average["rating_pred"] = test_item_average["rating_pred"].fillna(global_mean)

# rmse μεταξύ train και test data
mse = mean_squared_error(test_item_average["rating"], test_item_average["rating_pred"])
rmse_item_average = np.sqrt(mse)

print("RMSE (Item Average):", rmse_item_average)
```

- Για το user average η πρόβλεψη για κάθε χρήστη είναι ο μέσος όρος των ratings από το training set.

- Για το item average είχαμε ταινίες που δεν υπήρχαν στο training set, γι αυτό για να γίνει η πρόβλεψη κάναμε αντικατάσταση των NaN γραμμών με το μέσο όρο των rating.

## Βήμα 2:

```python
from scipy.sparse import csr_matrix   # compressed sparse row matrix
from scipy.sparse.linalg import svds  # svd sparse
import matplotlib.pyplot as plt

# υπολογίζουμε τις διαστάσεις του πίνακα R (+1 διότι ids ξεκινάνε από 0)
users_count  = int(train_dataframe["uid"].max() + 1)
movies_count = int(train_dataframe["mid"].max() + 1)


# φτίαχνουμε τον R (αραιός πίνακας)
R = csr_matrix(
    (train_dataframe["rating"].astype(float).values,
     (train_dataframe["uid"].values, train_dataframe["mid"].values)),
    shape=(users_count, movies_count)
)

# υπολογίζουμε τους 3 πίνακες του svd (sparse)
K = 20
U, s, Vt = svds(R, k=K)


# αναστρέφουμε τη σειρά των singular values σε φθίνουσα
idx = np.argsort(s)[::-1]
s = s[idx]
U = U[:, idx]
Vt = Vt[idx, :]


# παίρνουμε τις τιμές του αρχικού dataframe ώστε να μην ξαναδημιουργήσουμε
# τον πίνακα R, υπολογίζοντας μόνο τις συγκεκριμένες (k) τιμές από αυτόν
u_test = test_dataframe["uid"].values
m_test = test_dataframe["mid"].values
r_test = test_dataframe["rating"].astype(float).values

# empty list to store the errors for different K values
rmse_list = []

# για κάθε k
for i in range (1, K+1):

    # υπολογίζουμε μόνο τις συγκεκριμένες τιμές του R_k
    predictions = np.sum(
      U[u_test, :i] * s[:i] * Vt[:i, m_test].T,
      axis=1
    )

    predictions = np.clip(predictions, 0, 5)  # clip τις τιμές <0 και >5

    # υπολογισμός rmse
    mse = mean_squared_error(r_test, predictions)
    rmse = np.sqrt(mse)

    rmse_list.append(rmse)

    print(f"K={i:2d}  RMSE={rmse:.6f}")

# κρατάμε το καλύτερο k - αυτό με μικρότερο σφάλμα
best_K = int(np.argmin(rmse_list) + 1)
best_rmse = rmse_list[best_K - 1]  # κρατάμε και το καλύτερο error

# plots
plt.figure()
plt.plot(range(1, K + 1), rmse_list, marker="o")
plt.xlabel("K")
plt.ylabel("RMSE")
plt.title("SVD: RMSE vs K")
#plt.grid(True)
plt.show()

print(f"\nBest K = {best_K} with RMSE = {best_rmse:.6f}")


```

```python
def user_mean(R, u):
    user_ratings = R[u].data  # παίρνουμε τα στοιχεία του R, μη μηδενικά εφόσον είναι αραιός πίνακας
    if len(user_ratings) == 0:
        return 0.0
    return user_ratings.mean()  # μέσος όρος βαθμολογιών χρήστη


def item_users(R, m):
    # παίρνουμε όλους τους users που έχουν κάνει rate την ταινία m
    # nonzero() επιστρέφει πλειάδα τύπου (array[1,2], )
    # παίρνουμε το 1ο στοιχείο, δηλαδή το array[1,2]
    # δηλαδή users με user id 1 και 2 έχουν κάνει rate μια ταινία m
    return R[:, m].nonzero()[0]


def similar_users(R, u, m_users, k):
    Ru = R[u]  # τα ratings του u
    # υπολογίζουμε την L2 νόρμα (.A1 μετατρέπει σε 1D πίνακα)
    norms = np.sqrt(R.power(2).sum(axis=1)).A1

    # αν ο u δεν έχει ratings δεν έχουμε similarity
    if norms[u] == 0:
        return np.array([], dtype=int), np.array([])

    # αν ο u βρίσκεται στο m_users (για να μην υπολογίσει similarity με τον εαυτό του)
    m_users = m_users[m_users != u]

    # κρατάμε μόνο χρήστες με μη μηδενική νόρμα
    valid = norms[m_users] > 0
    m_users = m_users[valid]

    # αν δεν έχουμε καθόλου χρήστες με μη μηδενική νόρμα
    if len(m_users) == 0:
        # επιστρέφουμε κενό array με γείτονες και κενό array για ομοιότηες
        # δεν έχουμε γείτονες και ομοιότητες
        return np.array([], dtype=int), np.array([])


    # υπολογισμός cosine similarity
    # dot products μόνο με τους χρήστες που είναι στο m_users
    similarities = R[m_users].dot(Ru.T).A1  # αριθμιτής
    similarities = similarities / (norms[m_users] * norms[u])  # διαίρεση με το γίνομενο από τις 2 νόρμες

    # αντιστροφή σειράς για να πάρουμε τα indexes από
    # μεγάλο σε μικρό similarity και να κρατήσουμε τα top k πιό similar
    idx = np.argsort(similarities)[::-1][:k]
    return m_users[idx], similarities[idx]


def compute_score(k_ratings, k_similarities):
    denominator = np.sum(k_similarities)  # παρονομαστής
    if denominator == 0:
        return 0.0
    return np.dot(k_similarities, k_ratings) / denominator  # αριθμιτής / παρονομαστής

```
